```python
from transformers import AutoTokenizer
```

```python
MODEL = f"cardiffnlp/twitter-roberta-base-sentiment-latest"
tokenizer = AutoTokenizer.from_pretrained(MODEL)
```

```python
from transformers import pipeline
sentiment_task = pipeline("sentiment-analysis", model=MODEL, tokenizer=tokenizer)
```

```python
sentiment_task('I love Hugging face')
```

```python
#Create a table of reviews
import pandas as pd

# Define review data as a list of dictionaries
data = [
    {"Restaurant": "Little Italy", "Rating": 5, "Review": "Delicious food, great service!"},
    {"Restaurant": "Burger Barn", "Rating": 3, "Review": "Average burger, slow service."},
    {"Restaurant": "Spicy Curry", "Rating": 4, "Review": "Excellent curry, a bit pricey."},
    {"Restaurant": "The French Bistro", "Rating": 1, "Rating": "Waited an hour for food, never came."},
    {"Restaurant": "Cozy Cafe", "Rating": 4, "Review": "Nice atmosphere, coffee a bit weak."},
    {"Restaurant": "Yum Yum Noodles", "Rating": 5, "Review": "Fantastic noodles, friendly staff!"},
]

# Create a DataFrame from the data
df = pd.DataFrame(data)

df
```

```python
df['Sentiment'] = 'NaN'
```

```python
for i in range(len(df['Review'])):
  df['Sentiment'][i] = sentiment_task(str(df['Review'][i]))
```

```python
df
```

```python
df['label'] = df['Sentiment'].apply(lambda x:x[0]['label'])
```
